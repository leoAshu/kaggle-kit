```python
import pandas as pd

import torch
from torch import nn

from kagglekit import (
    Dataset,
    LinearClassifier,
    Trainer,
    plot_histories,
    summarize_histories,
)
```

## Load Full Train Set

```python
data = torch.load('../data/train_full.pt')

X = data['X']

Y = data['Y'].float().reshape(-1, 1)

X.shape, Y.shape
```

```python
dataset = Dataset(X, Y, batch_size=32)
```

### Training

```python
torch.manual_seed(0)

model = LinearClassifier(in_features=X.shape[1])

loss_fn = nn.BCEWithLogitsLoss()
optim = torch.optim.Adam(model.parameters(), lr=0.01)

trainer = Trainer(loss_fn, optim, max_epochs=17)

history = trainer.fit(model, dataset)
```

### Load Test Set

```python
test = torch.load('../data/test.pt')

X = test['X']
ids = test['PassengerId']

X.shape, ids.shape
```

#### Prediction

```python
model.eval()

preds = model.predict(X.to(trainer.device)).cpu()

preds.shape
```

#### Submission

```python
submission = pd.DataFrame({
    'PassengerId': ids,
    'Survived': preds,
})
submission.to_csv('../submissions/linear_adam_0.01_17.csv', index=False)

submission.head()
```
